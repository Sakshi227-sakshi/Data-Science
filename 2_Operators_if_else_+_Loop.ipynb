```python
#While loop example -> program t o print the table
#Program -> Sum of all digits of a given number
#Program -> keep accepting number from user till he/she  enters a 0 an dthan find the avg
```

```python
number = int(input('enter the number'))

i = 1

while i<11:
  print(number,'*',i,'=',number * i)
  i += 1
```

```python
#while loop with else

x = 1

while x < 3:
  print(x)
  x += 1

else:
  print('limit crossed')
```
