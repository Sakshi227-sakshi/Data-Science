#******List Comprehension
list comprehension provides a concise way af creating lists.

* newlist = [expressions for item in iterable if condition == True]

## Advantage of list Comprehension ----
    1. more time- efficient and space-efficient than loops.
    2. Requier fewer Linesvof Code.
    3. Transforms iterative statement innto s formula.

```python
#Add 1 to 10 number to a list

L = []

for i in range(1,11):
  L.append(i)
print(L)

L = [i for i in range(1,11)]
print(L)
```

```python
#scalar multiplication on a vector
v = [2,3,4]
s = -3
#-6,-9,-12
[s*i for i in v]
```

```python
#Add Squares
L = [1,2,3,4,5]
[i**2 for i in L]
```

```python
#Q1. Print all number divisible by 5 in the range of 1 to 50

[i for i in range(1,51) if i%5 == 0]
```

```python
#Q2. find language which start with letter p

languages = ['java0','python','php','c','javascript']
[language for language in languages if language.startswith('p')]
```

```python
#Q3. Nested if with List Comprehension
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']

#add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'

[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
```

```python
#Print a (3,3) matrix using list comprehension  -> Netsted List Comprehansion

[[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
#cartesian product -> List Comprehension on 2 list together
L1 = [1,2,3,4]
L2 = [5,6,7,8]
[[i*j for i in L1 for j in L2]]
```

#**2 Ways to traverse a list

```python
#itemwise
L = [1,2,3,4]
for i in L:
  print(i)
```

```python
#indexwise
L = [1,2,3,4]
for i in range(0,len(L)):
  print(L[i])
```

** Zip** -> The zip() function return a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together,and then the second item in each passed iterator are paired together.

*if the passed iterators have different length ,the iterator with the least item decides the length of the new iterator.*

```python
#write a program to add item of 2 list indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

list(zip(L1,L2))
[i+j for i,j in zip(L1,L2)]
```
