# Q1. Sequence sum     1/1!+ 2/2!+3/3!

```python
n = int (input('enter n '))

result = 0
fact = 1

for i in range(1,n+1):
  fact =  fact * i
  result = result + i/fact

print(result)
```

```python
#Q count the frequency of a perticulaer character in a provide string.

#Eg'hello hoe are you' is the string, the frequncy of h in this string is 2'

s = input('enter the emails ')
term =input('what would like to search for ')

counter = 0
for i in s:
  if i == term:
    counter += 1
print('frequency',counter)
```

```python
#Q write a program which can remove a particuler character from a string.

s = input('enter the string ')
term =input('what would like to remove ')

result =  ''

for i in s:
  if i != term:
    result = result + i

print(result)
```

```python
#Q write a program that can check whether  a given string is palindrome or not .
#abba
#malyalam

s = input('enter the string ')
flag = True
for i in range(len(s)//2):
  if s[i] != s[len(s)- i-1]:
    flag = False
    break
if flag:
    print('palindrom')
else:
    print('Not a palindrom')
```

```python
#Q write a program to count the number odf words in a string without split...

s = input('enter the string ')
L = []
temp =''
for i in s:

  if i != ' ':
    temp = temp+i
  else:
    L.append(temp)
    temp =''
L.append(temp)
print(L)
```

```python
#Q write a pythom program to  convert a string to title case without using the title()

s = input('enter the string ')

L = []
for i in s.split():
  #print(i)
  L.append(i[0].upper()+ i[1:].lower())
#print(L)
print(" ".join(L))
```

```python
#Q write a program thar can convert an integer to string

number = int(input('enter the number :'))
digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result
  number =  number//10

print(result)
```
